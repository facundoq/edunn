# Mean or average error

While the SquaredError layer lets us compute the errors of each example, to obtain a measure of the error over a batch or set of examples, we need to compute the average of these errors. Since this computation is independent of the error function, we can encapsulate it in its own class. Implement the `forward` method of the mean error class.

Note: the error function is often also called the _loss_, to distinguish it from the mean error

```python
y = np.array([[2,-2],
             [-4,4]])
y_true = np.array([[3,3],
             [-5,2]])


layer=nn.MeanError(nn.SquaredError())
E=15.5
utils.check_same_float(E,layer.forward(y,y_true),title="mean error")
```
