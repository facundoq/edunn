# Softmax Layer


For classification problems, it is useful to have a model that can output probability distributions. This way, for a problem with `C` classes, the model can output a vector of `C` elements `f(x)=y` where each `y_i` is a value between 0 and 1 that indicates the probability that the example `x` belongs to class `i`. Also, since `y` is a distribution, we have that $\sum_{i=1}^C y_i =1$,

A linear regression model can output a vector of `C` elements with the _scores_ for each class, but those values will lie in the range $(-\infty, +\infty)$, so they could never satisfy the properties of a probability distribution we mentioned above.

In this exercise you must implement the `Softmax` layer, which does exactly that: given a vector `x` of `C` per-class scores, it converts it into a vector `y` of per-class probabilities. To do so, it implements the Softmax function, with $y =(y_1,y_2,...,y_C) = Softmax((x_1,x_2,...,x_C)) = Softmax(x)$, where each $y_i$ is the probability of class $i$.  So, for example, given the class scores `[-5,100,100]`, the `Softmax` function will produce the probabilities `[0,0.5,0.5]`.




# Forward method

The `forward` method uses the following formula for `y`:

$$y= 
\frac{[e^{x_1},...,e^{x_c}]}{e^{x_1}+...+e^{x_c}} = 
\frac{[e^{x_1},...,e^{x_c}]}{N}$$

Or, seen element by element, each value of $y$ is defined as:
$$y_i(x) =  \frac{e^{x_i}}{e^{x_1}+...+e^{x_c}} $$

Here we use the exponential function ($e^x$) to convert each score of the vector `x` from the range $(-\infty, +\infty)$ to the range $(0, +\infty)$, since that is the image of the exponential function. 

Furthermore, $e^x$ is monotonically increasing in $x$, so larger values of $x$ give larger values of $e^x$; therefore, if a score is high, so is the probability, and vice versa. 

Now, on top of that, each element is divided by the value $N$, which serves to normalize the values, giving us:
1. Values between 0 and 1
2. That the values sum to 1

That is, the axioms of a probability distribution as we mentioned above

Implement the `forward` method of the `Softmax` class.


```python
x = np.array([[0,0,100],
             [0,100,0.0],
             [100,100,0.0],
             [50,50,0.0],
             [1,1,1],],dtype=float)

layer=nn.Softmax()
y = np.array([[ 0, 0,  1],
               [0, 1, 0.0],
             [0.5, 0.5, 0.0],
              [0.5, 0.5, 0.0],
             [1/3,1/3,1/3]],dtype=float)

utils.check_same(y,layer.forward(x))
```
