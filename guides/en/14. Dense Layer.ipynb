# Dense or FullyConnected Layer

The `Linear`, `Bias` and activation layers (`Sigmoid`, `ReLU`, `TanH` and others) are usually used together in the form `dense(x) = activation(w*x+b)`, where `activation` is one of these activation functions. This layer is commonly called `FullyConnected` or, as we will call it here, `Dense`; the name comes from the fact that each output of the layer depends on *all* of the inputs.

In this exercise you must implement the `Dense` layer, but using the `Linear`, `Bias` and activation layers directly, without copying their code.




# Creation and Initialization

The `Dense` layer should have a parameter vector `w` and another vector `b`, each one with its own initializer. In addition, it should have an activation function.

But to implement it, we are going to use 3 *internal layers*: `Linear`, `Bias` and the activation one, which we will call `Activation`. For convenience, we will also let you specify the activation with a string such as `relu`, `sigmoid` or `tanh`. In this case, we help you out by already defining the `__init__` constructor, which assigns the corresponding *internal layer* objects to the `self.linear`, `self.bias` and `self.activation` instance variables and lets you specify their initializers.

We also help you out by implementing `get_parameters`, which combines the parameter dictionary of each sublayer into a single large dictionary of `Dense` gradients.

We recommend that you study the code of these two methods (`__init__` and `get_parameters`) to see how they work. They will help you afterwards when implementing the `forward` and `backward` of `Dense`.



```python
# We create a Dense layer with 2 input values and 3 output values
# and ReLU activation
# The linear layer is initialized randomly
# while the bias layer is initialized to 0

input_dimension=2
output_dimension=3
activation="relu"
dense1=nn.Dense(input_dimension,output_dimension,
                 activation_name="relu",
                 linear_initializer=nn.initializers.RandomNormal(),
                 bias_initializer=nn.initializers.Constant(0),
                 )
print(f"Layer name: {dense1.name}")
print(f"Parameter w of the layer: {dense1.get_parameters()['w']}")
print("(it must change every time you re-run this cell)")
print(f"Parameter b of the layer: {dense1.get_parameters()['b']}")
print("(it must always be 0)")
print()
```

# The `forward` method


Now that we know how to create and initialize `Dense` layer objects, we begin with the `forward` method, which you can find in the file `dense.py` inside the `edunn/models` folder.

To implement the forward pass, you must take the input `x` and use it to call the `forward` of the internal layers of type `Linear`, `Bias` and `Activation`.

To verify that the `forward` implementation is correct, we use the `Constant` initializer both times, but afterwards the layer keeps using a random initializer such as `RandomNormal` by default.


```python
x = np.array([[3,-7],
             [-3,7]])

w = np.array([[2, 3, 4],[4,5,6]])
b = np.array([1,2,3])
linear_initializer = nn.initializers.Constant(w)
bias_initializer = nn.initializers.Constant(b)
layer=nn.Dense(2,3,linear_initializer=linear_initializer,bias_initializer=bias_initializer)
y = np.array([[-21, -24, -27],
              [ 23, 28,  33]])

nn.utils.check_same(y,layer.forward(x))

linear_initializer = nn.initializers.Constant(-w)
bias_initializer = nn.initializers.Constant(-b)
layer=nn.Dense(2,3,linear_initializer=linear_initializer,bias_initializer=bias_initializer)
nn.utils.check_same(-y,layer.forward(x))
```

# The `backward` method


To implement `backward`, you must also call the `backward` of the `self.linear`, `self.bias` and `self.activation` variables, in the correct order and manner. Hint: it is the reverse of the `forward` one.

In this case, we also help you out by combining the gradient dictionary of each layer into a single large dictionary of `Dense` gradients, using the `**dict` operator that unpacks a dictionary and `{**dict1, **dict2}` that puts them back together.


```python
samples = 100
batch_size=2
features_in=3
features_out=5
input_shape=(batch_size,features_in)

# Test derivatives of a Dense layer with random values for `w`
layer=nn.Dense(features_in,features_out,activation_name='relu')

nn.utils.check_gradient.common_layer(layer,input_shape,samples=samples)    
```
