# Flatten Layer

The flattening operation (Flatten) in a convolutional neural network is quite simple. In the `forward` method, you simply need to reshape the input tensor into a vector, preserving the batch dimension. In the `backward` method, you need to reshape that vector back into the original shape of the input tensor. Implement both methods.

```python
np.random.seed(123)

x = np.random.rand(2,3,5,5)

layer=nn.Flatten()
y=layer.forward(x)
```

```python
# Define the gradient of the output
g = np.random.rand(*y.shape)

# Propagate the gradient backward through the convolution
layer_grad = layer.backward(g)
layer_grad
```

## Checks with PyTorch

```python
import torch
import torch.nn as tnn

x = torch.from_numpy(x).to(torch.double)
x.requires_grad = True

flatten = tnn.Flatten()
y_torch = flatten(x)
y_torch
```

```python
utils.check_same(y_torch.detach().numpy(),y)
```

```python
# Define the gradient of the output
g = torch.from_numpy(g).to(torch.double)

# Propagate the gradient backward through the convolution
y_torch.backward(g)

# Print the gradient of the input image
print("Input gradient (δE/δx):")
print(x.grad, x.grad.shape)
```

```python
utils.check_same(x.grad.numpy(),layer_grad[0])
```
