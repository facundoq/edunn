# Linear Regression Model: `forward`

A Linear Regression model is built by first applying a linear layer $f(x)=xw$ and then a bias layer $f(x)=x+b$, finally obtaining the function $f(x)=xw+b$. Now, instead of viewing $f(x)$ as $xw+b$, we can view it as `x -> Linear -> Bias -> y`. That is, as a succession of layers, each one transforming the input `x` until we obtain the output `y`.

In terms of code, the `forward` method of a `LinearRegression` model is the composition of the `forward` functions of the `Linear` and `Bias` layers:
````python
y_linear = linear.forward(x)
y = bias.forward(y_linear)
````

Implement the `forward` method of the `LinearRegression` model in the file `edunn/models/linear_regression.py`:

`````python
x = np.array([[3,-7],
             [-3,7]])

w = np.array([[2, 3, 4],[4,5,6]])
b = np.array([1,2,3])
linear_initializer = nn.initializers.Constant(w)
bias_initializer = nn.initializers.Constant(b)
layer=nn.LinearRegression(2,3,linear_initializer=linear_initializer,bias_initializer=bias_initializer)
y = np.array([[-21, -24, -27],
              [ 23, 28,  33]])

utils.check_same(y,layer.forward(x))

linear_initializer = nn.initializers.Constant(-w)
bias_initializer = nn.initializers.Constant(-b)
layer=nn.LinearRegression(2,3,linear_initializer=linear_initializer,bias_initializer=bias_initializer)
utils.check_same(-y,layer.forward(x))
`````

# Linear Regression Model: `backward`

The `backward` method of a `LinearRegression` model is the *reverse* composition of the `backward` functions of the `Linear` and `Bias` layers.

That is, the derivative of the error with respect to the model's output first passes through `Bias`, which computes the derivative with respect to its parameters and returns it in `δEδbias` (in this case the only parameter is $b$). We then have `δEδx_bias`, the derivative of the error with respect to the output of the `Linear` layer. Therefore, we can do the same as in the `forward`, but in reverse order.

This is the first (simple) example of applying the *backpropagation* algorithm! This time, with only two models/layers. Later on we will generalize it with the `Sequential` model.

In this case, we also help you out by combining each layer's gradient dictionary into a single big dictionary of `LinearRegression` gradients, using the `**` operator that unpacks a dictionary, together with `{**dict1, **dict2}` which merges them back together.



`````python
samples = 100
batch_size=2
din=3 # input dimension
dout=5 # output dimension
input_shape=(batch_size,din)

# Check the derivatives of a Linear Regression model
# with random values for `w`, `b`, and `x`, the input
layer=nn.LinearRegression(din,dout)

utils.check_gradient.common_layer(layer,input_shape,samples=samples)    
`````
