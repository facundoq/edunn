# The backward method

Besides computing the layer's output, the layer must also be able to backpropagate the network's error gradient. To do that, you must implement the `backward` method, which receives $\frac{δE}{δy}$, that is, the partial derivatives of the error with respect to this layer's output (its gradient), and returns $\frac{δE}{δx}$, the partial derivatives of the error with respect to this layer's inputs.

For the `MultiplyConstant` layer, computing the gradient is easy, since given:

$
y(x_1,x_2,...,x_n)= (x_1*C,x_2*C,...,x_n*C)
$

Then

$y_i(x)=x_i*C$

And so

$\frac{δE}{δx_i} = \frac{δE}{δy} * \frac{δy}{δx_i}$

Since $y_i$ depends only on $x_i$, we can rewrite the above as

$ \frac{δE}{δx_i} = \frac{δE}{δy} * \frac{δy}{δx_i} = \frac{δE}{δy_i} * \frac{δy_i}{δx_i} $

Given that $y_i(x)=x_i*C$, then $\frac{δy_i}{δx_i} = C$ and therefore:
$ \frac{δE}{δx_i} = \frac{δE}{δy} * \frac{δy}{δx_i} = \frac{δE}{δy_i} * C $

Writing this in vector form for the vector x:

$ \frac{δE}{δx} = [ \frac{δE}{δy_1} *C, \frac{δE}{δy_2} *C, ..., \frac{δE}{δy_n}*C ] = \frac{δE}{δy}*C $

So the layer simply propagates the gradients coming from the next layer, but multiplied by C.
